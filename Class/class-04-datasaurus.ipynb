```python
x = np.linspace(-50, 0, 200)
datasets = { "Lineal": 2*x + 1, "Cuadratica": x**2, "Seno": np.sin(x)}
```

```python
for name, y in datasets.items():
    df = pd.DataFrame({"x": x, "y": y})
    pearson = df.corr().loc["x", "y"]
    spearman = df.corr(method="spearman").loc["x", "y"]
    plt.figure()
    plt.scatter(df["x"], df["y"], s=10)
    plt.title(f"P={pearson:.2f}, S={spearman:.2f}")
    plt.show()
```

```python
def M_D_V_C(x,y):
    print(f"la media de {x} es: ", np.mean(x))
    print(f"La desviasión estándar de {x} es " , np.std(x))
    print(f"La varianza de {x} es ", np.var(x))
    print(f"la media de {y} es: ", np.mean(y))
    print(f"La desviasión estándar de {y} es " , np.std(y))
    print(f"La varianza de {y} es ", np.var(y))
    print("el coeficiente de correlación de pearson es " , np.corrcoef(x,y)[0,1])
    plt.scatter(x,y)
```
