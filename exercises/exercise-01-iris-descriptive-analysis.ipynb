```python
def histograma (x):
    
    longitud_sepalo = []
    ancho_sepalo = []
    longitud_petalo = []
    ancho_petalo = []

    with open(r"C:\Users\Piloto 01\Documents\UNAM universidad\9vo Semestre\Ciencia de datos\Prácticar fuera de clase\iris\iris.csv") as f:

        for cols in csv.reader(f):

            if cols[4] == x:

                longitud_sepalo.append(float(cols[0]))
                ancho_sepalo.append(float(cols[1]))
                longitud_petalo.append(float(cols[2]))
                ancho_petalo.append(float(cols[3]))

    plt.hist(longitud_sepalo, density=True, alpha=0.2,
         color="blue", label="Longitud del sépalo")

    plt.hist(ancho_sepalo, density=True, alpha=0.2,
         color="red", label="Ancho del sépalo")

    plt.hist(longitud_petalo, density=True, alpha=0.2,
         color="yellow", label="Longitud del pétalo")

    plt.hist(ancho_petalo, density=True, alpha=0.2,
         color="black", label="Ancho del pétalo")
    plt.title(f"Histograma asociado {x}")

    plt.legend()

    plt.show()

    plt.show()
```
