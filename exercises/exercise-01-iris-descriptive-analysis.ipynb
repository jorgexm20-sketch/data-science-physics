### 1. Diagrama de caja y bigotes (Boxplot)

El **boxplot** permite observar la mediana, los cuartiles, la dispersión de los datos y la posible presencia de valores atípicos.

```python

```

### 2. Diagrama de violín (Violin plot)

El **violin plot** combina información similar a la de un boxplot con una representación de la densidad de los datos. Permite observar la forma de la distribución y las regiones donde se concentra una mayor cantidad de observaciones.

```python
sns.violinplot(
    data=DATAFRAME,
    x='COLUMNA_CATEGORICA',
    y='COLUMNA_NUMERICA'
)

plt.show()
```

---

### 3. Gráfico de puntos (Stripplot)

El **stripplot** representa individualmente cada observación del conjunto de datos. Es útil para identificar dónde se concentran los valores y comparar visualmente diferentes grupos.

```python
sns.stripplot(
    data=DATAFRAME,
    x='COLUMNA_CATEGORICA',
    y='COLUMNA_NUMERICA',
    jitter=True
)

plt.show()
```

El parámetro `jitter=True` introduce un pequeño desplazamiento horizontal para evitar que los puntos con valores similares se superpongan completamente.

---

### 4. Diagrama de dispersión (Scatter plot)

El **scatter plot** permite estudiar la relación entre dos variables numéricas. Cada punto representa una observación y su posición depende de los valores de las dos variables seleccionadas.

```python
sns.scatterplot(
    data=DATAFRAME,
    x='VARIABLE_X',
    y='VARIABLE_Y',
    hue='COLUMNA_CATEGORICA'
)

plt.show()
```

El parámetro `hue` permite diferenciar las observaciones de acuerdo con una variable categórica.

---

### 5. Matriz de gráficos de dispersión (Pairplot)

El **pairplot** genera automáticamente gráficos de dispersión entre diferentes pares de variables numéricas. También muestra la distribución individual de cada variable en la diagonal principal.

```python
sns.pairplot(
    DATAFRAME,
    hue='COLUMNA_CATEGORICA'
)

plt.show()
```
